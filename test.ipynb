```python
from transformers import AutoTokenizer
from logging import getLogger
import torch
logger = getLogger()
logger.setLevel("INFO")

context_length = 100

input_text = ["This story is", "I am a", "This script is", "Mercury "]
tokenizer = AutoTokenizer.from_pretrained("microsoft/phi-2")
input_ids = tokenizer(input_text, return_tensors="pt")["input_ids"].to("cuda:1")
full_sentence = input_ids.clone().cpu()
head_count = torch.zeros(5, dtype=torch.int32)
for i in range(1000):
    output = model(input_ids)
    head_count += torch.bincount(output.head_indices, minlength=5).cpu()
    if i % 50 == 0:
        print(f"Step {i} : head count {head_count}, input_ids {input_ids.shape}")
    next_token_logits = output.logits[:, -1, :]
    next_token_id = torch.multinomial(torch.nn.functional.softmax(next_token_logits, dim=-1), num_samples=1).squeeze()
    full_sentence = torch.cat([full_sentence, next_token_id.unsqueeze(-1).cpu()], dim=-1)
    input_ids = torch.cat([input_ids, next_token_id.unsqueeze(-1)], dim=-1)
    if input_ids.shape[-1] > context_length:
        input_ids = input_ids[:, -context_length:]
for sentence in full_sentence:
    print(tokenizer.decode(sentence))
with open(f"head_usage_phi-2_{model_penalty}.txt", "w") as f:
    f.write(str(head_count.tolist()))
```

```python
from transformers import AutoModel
from src.BranchyModelConfig import BranchyModelConfig
from src.BranchyModel import BranchyCausalModel
import torch

device = "cpu"
model_penalty = "0.9"
model_conf = BranchyModelConfig.from_pretrained(f"valcore/branchy_phi-2_{model_penalty}")
model_conf.device = device
model = BranchyCausalModel(model_conf).to(device)
pretrained_backbone = torch.load("base_model.bin", map_location=device)
pretrained_head = torch.load("head_model_phi2_penaltyv2_0.9.bin", map_location=device)
model.model.load_state_dict({**pretrained_backbone, **pretrained_head})
```

```python
heads_place = model.config.branch_locations
count = [0] * 5
print(heads_place)
for name, param in model.named_parameters():
    if name.startswith("model.model.model.layers."):
        name = name.split(".")
        layer = int(name[4])
        if layer < heads_place[0]:
            count[0] += param.numel()
        elif layer < heads_place[1] and layer >= heads_place[0]:
            count[1] += param.numel()
        elif layer < heads_place[2] and layer >= heads_place[1]:
            count[2] += param.numel()
        elif layer < heads_place[3] and layer >= heads_place[2]:
            count[3] += param.numel()
        elif layer >= heads_place[3]:
            count[4] += param.numel()
    elif name.startswith("model.model.model.embed"):
        count[0] += param.numel()
    elif name.startswith("model.model.model.final"):
        count[4] += param.numel()
    elif name.startswith("model.model.lm_head"):
        count[4] += param.numel()
    elif name.startswith("model.branches.0"):
        count[0] += param.numel()
    elif name.startswith("model.branches.1"):
        count[1] += param.numel()
    elif name.startswith("model.branches.2"):
        count[2] += param.numel()
    elif name.startswith("model.branches.3"):
        count[3] += param.numel()
    else:
        print(name)
print(count)
```

```python
import matplotlib.pyplot as plt
import numpy as np

cumsum_count = np.cumsum(count)

# bar plot of cumsum of count
fig, ax = plt.subplots()
ax.bar(np.arange(5), cumsum_count)
ax.set_xticks(np.arange(5))
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import glob

usage_list = []
files = sorted(glob.glob("head_usage_phi-2_*.txt"))
for usage_file in files:
    penalty = usage_file.split("_")[-1].split(".txt")[0]
    with open(usage_file, "r") as f:
        usage = f.read()
    usage = eval(usage)
    usage_list.append(np.array(usage)/sum(usage))
print(usage_list)

lambdas = [0.4, 0.5, 0.6, 0.7, 0.8, 0.9]

# Create a DataFrame from the data
df = pd.DataFrame(usage_list, columns=['Head 0', 'Head 1', 'Head 2', 'Head 3', 'Main model'])
df['lambda'] = lambdas

# Melt the DataFrame to long format
df_melted = df.melt(id_vars='lambda', var_name='Category', value_name='Value')

# Create the bar plot
plt.figure(figsize=(10, 6))
sns.barplot(x='lambda', y='Value', hue='Category', data=df_melted, palette='viridis')

# Add title and labels
plt.title('% of data outputed by each head')
plt.xlabel('Epsilon')
plt.ylabel('Value')
plt.legend(title='Category')

# Show the plot
plt.show()
```

```python
import glob
import seaborn as sns
import pandas as pd

base_model_param_used = cumsum_count[-1] * 4000
usage_dict = []
files = sorted(glob.glob("head_usage_phi-2_*.txt"))
print(files)
for usage_file in files:
    penalty = usage_file.split("_")[-1].split(".txt")[0]
    with open(usage_file, "r") as f:
        usage = f.read()
    usage = eval(usage)
    params_used = usage * cumsum_count
    usage_dict.append(base_model_param_used / sum(params_used))
usage_dict.append(1)
print(len(usage_dict))
# make bar plot with x being 0.4, 0.5, ... 1
x = [0.4, 0.5, 0.6, 0.7, 0.8, 0.9, 1]

data = pd.DataFrame({'Lambda': x, 'Speedup': usage_dict})

# Plotting using Seaborn
plt.figure(figsize=(8, 6))
sns.lineplot(data=data, x='Lambda', y='Speedup', marker='o', palette="viridis")
plt.gca().invert_xaxis()

# Adding titles and labels
plt.title('Speedup vs Epsilon', fontsize=16)
plt.xlabel('Epsilon (ε)', fontsize=14)
plt.ylabel('Speedup (x)', fontsize=14)
plt.grid(True)
```
